# Missing dates investigation 

```python
# before using Lad_7 in forcasting, lets understand data frequency

df_cleaned['Date_diff'] = df_cleaned['Date'].diff()

print( df_cleaned[['Date', 'Date_diff']].head(20))
```

```python
# Let's find out how many dates are missing.

expected_date = pd.date_range(
    start= df_cleaned['Date'].min(),
    end= df_cleaned['Date'].max(),
    freq= 'D'
)

missing_date = expected_date.difference(
    df_cleaned['Date']
)

print('Expected date ', len(expected_date))
print('Actual date ', df_cleaned['Date'].nunique())
print('missing date ', len(missing_date))
```

```python
# converts missing dates into dataframe

missing_df = pd.DataFrame({
    'Date' : missing_date
})
# get the weekly name 
missing_df['Day'] = missing_df['Date'].dt.day_name()
# count missing dates by weekday
missing_by_day = missing_df['Day'].value_counts()

print(missing_by_day)
```

```python
# missing dates by year and month
# add year and month

missing_df['Year'] = missing_df['Date'].dt.year
missing_df['Month'] = missing_df['Date'].dt.month

# count of missing dates by year
missing_by_year = missing_df['Year'].value_counts().sort_index()
print('Missing dates by Year :')
print(missing_by_year)

# count of missing dates by month
missing_by_month = missing_df['Month'].value_counts().sort_index()
print('\nMissing dates by month :')
print(missing_by_month)
```

```python
# consecutive missing date
# sort missing dates
missing_date_sort = pd.Series(missing_date).sort_values().reset_index(drop=True)

# calculate the gap between consecutive missing date
gap = missing_date_sort.diff().dt.days
print(gap.value_counts().sort_index())
```

```python
# which weekdays values have actualy observed
df_cleaned['Day'] = df_cleaned['Date'].dt.day_name()
actual_by_day = df_cleaned['Day'].value_counts()
print(actual_by_day)
```
